```python
import os
from datasets import load_dataset
import pandas as pd
```

```python
os.makedirs('data/raw', exist_ok=True)
```

```python
dataset_id = "KFUPM-JRCAI/arabic-generated-abstracts"
print("جاري تحميل البيانات من Hugging Face...")
dataset = load_dataset(dataset_id)
```

```python
for split_name in dataset.keys():
    df = pd.DataFrame(dataset[split_name])
    file_path = f"data/raw/arabic_abstracts_{split_name}.csv"
    df.to_csv(file_path, index=False, encoding='utf-8-sig ترميز عربي')
    print(f"تم حفظ قسم '{split_name}' بنجاح في: {file_path} (عدد العينات: {len(df)})")
```
